### Ejercicio 4.1: Análisis de datos médicos - Columnas

Carga el dataset de colesterol y realiza las siguientes operaciones con columnas:
- Carga el archivo 'colesterol.csv' desde el repositorio de datos
- Selecciona solo las columnas numéricas del DataFrame
- Renombra las columnas usando nombres más descriptivos en español
- Crea una nueva columna 'IMC_Categoria' que clasifique el IMC (Índice de Masa Corporal)
- Crea una columna 'Rango_Colesterol' que indique si el colesterol es 'Alto' (>240), 'Moderado' (200-240) o 'Normal' (<200)



````python
import pandas as pd
# Ejercicio 4.1: Análisis de datos médicos - Columnas
col_path = '../../data/colesterol.csv'
df_col = pd.read_csv(col_path)

# Forzar columnas numéricas donde corresponda
for c in ['edad','peso','altura','colesterol']:
    if c in df_col.columns:
        df_col[c] = pd.to_numeric(df_col[c], errors='coerce')

# Renombrar columnas para mayor claridad
df_col = df_col.rename(columns={'peso':'peso_kg','altura':'altura_m','colesterol':'colesterol_mg'})

# Calcular IMC (manejar NaN)
df_col['IMC'] = df_col['peso_kg'] / (df_col['altura_m'] ** 2)

def imc_categoria(v):
    if pd.isna(v):
        return None
    if v < 18.5:
        return 'Bajo'
    elif v < 25:
        return 'Normal'
    elif v < 30:
        return 'Sobrepeso'
    else:
        return 'Obeso'

df_col['IMC_Categoria'] = df_col['IMC'].apply(imc_categoria)

def rango_col(x):
    if pd.isna(x):
        return None
    if x > 240:
        return 'Alto'
    elif x >= 200:
        return 'Moderado'
    else:
        return 'Normal'

df_col['Rango_Colesterol'] = df_col['colesterol_mg'].apply(rango_col)

print('Vista previa del dataset de colesterol:')
display(df_col.head())
print('\nColumnas numéricas:')
display(df_col.select_dtypes(include=['number']).head())
print('\nConteo por categoría de IMC:')
print(df_col['IMC_Categoria'].value_counts(dropna=False))
print('\nConteo por rango de colesterol:')
print(df_col['Rango_Colesterol'].value_counts(dropna=False))
````

### Ejercicio 4.2: Operaciones aritméticas entre columnas

Usando el DataFrame de ventas, realiza las siguientes operaciones:
- Crea una columna 'Descuento_10' con un descuento del 10% sobre el precio
- Crea una columna 'Precio_Final' que sea precio - descuento
- Calcula el total de ingresos por producto (precio × cantidad_vendida)
- Encuentra el producto con mayor margen de ganancia (asumiendo costo = precio * 0.6)

````python
import pandas as pd
# Ejercicio 4.2: Operaciones aritméticas entre columnas (ventas)
sales_path = '../../data/sales_data.parquet'
df_sales = pd.read_parquet(sales_path)

display(df_sales.head())

# Crear descuento del 10% y precio final
df_sales['Descuento_10'] = df_sales['priceeach'] * 0.10
df_sales['Precio_Final'] = df_sales['priceeach'] - df_sales['Descuento_10']

# Ingresos por fila (precio * cantidad)
df_sales['Ingresos_producto'] = df_sales['priceeach'] * df_sales['quantityordered']

# Total de ingresos por producto
ingresos_por_producto = df_sales.groupby('productcode')['Ingresos_producto'].sum().sort_values(ascending=False)

# Margen suponiendo costo = priceeach * 0.6
df_sales['margen_unit'] = df_sales['priceeach'] * 0.4
df_sales['margen_total'] = df_sales['margen_unit'] * df_sales['quantityordered']
producto_max_margen = df_sales.groupby('productcode')['margen_total'].sum().idxmax()

print('Vista previa ventas (head):')
display(df_sales.head())
print('\nTop 5 productos por ingresos:')
display(ingresos_por_producto.head())
print('\nProducto con mayor margen total:', producto_max_margen)
````

### Ejercicio 4.4: Análisis de datos de vivienda

Carga y analiza el dataset de housing para practicar operaciones avanzadas con columnas:
- Carga el archivo 'housing.csv' desde el repositorio de datos
- Crea una columna 'Precio_Por_M2' si existe información de precio y área
- Crea categorías de precio (Alto, Medio, Bajo) usando quantiles
- Calcula correlaciones entre las variables numéricas
- Identifica las columnas con mayor variabilidad
- Crea una columna binaria que indique si el valor está por encima de la mediana


````python
import pandas as pd
# Ejercicio 4.4: Análisis de housing
housing_path = '../../data/housing.csv'
df_house = pd.read_csv(housing_path)

print('Vista previa housing:')
display(df_house.head())

# Crear Precio_Por_M2 si hay información aproximada de área (usamos rooms per household como proxy)
if set(['median_house_value','total_rooms','households']).issubset(df_house.columns):
    df_house['rooms_per_household'] = df_house['total_rooms'] / df_house['households']
    df_house['Precio_Por_M2'] = df_house['median_house_value'] / df_house['rooms_per_household']
    print('Columna Precio_Por_M2 creada (proxy).')
else:
    print('No hay columnas suficientes para calcular Precio_Por_M2')

# Categorías de precio usando quantiles
q1 = df_house['median_house_value'].quantile(0.33)
q2 = df_house['median_house_value'].quantile(0.66)

def cat_precio(v):
    if v <= q1:
        return 'Bajo'
    elif v <= q2:
        return 'Medio'
    else:
        return 'Alto'

df_house['Categoria_Precio'] = df_house['median_house_value'].apply(cat_precio)

# Correlaciones y columnas con mayor variabilidad
corr = df_house.select_dtypes(include=['number']).corr()
variabilidad = df_house.select_dtypes(include=['number']).std().sort_values(ascending=False)
df_house['above_median'] = df_house['median_house_value'] > df_house['median_house_value'].median()

print('\nCorrelaciones (muestra):')
display(corr.iloc[:8,:8])

print('\nColumnas con mayor variabilidad:')
print(variabilidad.head())

print('\nConteo por categoría de precio:')
print(df_house['Categoria_Precio'].value_counts())

print('\nPorcentaje por encima de la mediana:')
print(df_house['above_median'].mean())
````

### Ejercicio 7.3: Funciones de agregación personalizadas

Para el DataFrame de ventas por región:
- Agrupa por región y aplica múltiples funciones: suma, promedio, mínimo, máximo
- Agrupa por región y encuentra la región con las ventas más altas y más bajas
- Agrupa por región y producto, calcula el rango de ventas (máximo - mínimo)

````python
import pandas as pd
# Ejercicio 7.3: Funciones de agregación personalizadas
# Usar el DataFrame de ventas del ejercicio anterior
ventas_region = {
    'region': ['Norte', 'Norte', 'Sur', 'Sur', 'Este', 'Este', 'Oeste', 'Oeste', 'Norte', 'Sur'],
    'producto': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'vendedor': ['Juan', 'Ana', 'Pedro', 'María', 'Luis', 'Carmen', 'José', 'Laura', 'Juan', 'Pedro'],
    'ventas': [1500, 2000, 1800, 2200, 1600, 1900, 1700, 2100, 1550, 1850],
    'unidades': [50, 60, 55, 65, 52, 58, 54, 62, 51, 56]
}
df_ventas = pd.DataFrame(ventas_region)

# Agrupación por región con múltiples funciones
print('Estadísticas por región (suma, promedio, mínimo, máximo):')
stats_por_region = df_ventas.groupby('region')['ventas'].agg(['sum', 'mean', 'min', 'max'])
display(stats_por_region)

# Encontrar la región con ventas más altas y más bajas
print('\nRegión con ventas más altas:')
region_max = stats_por_region['sum'].idxmax()
print(f'{region_max}: {stats_por_region.loc[region_max, "sum"]:.2f}')

print('\nRegión con ventas más bajas:')
region_min = stats_por_region['sum'].idxmin()
print(f'{region_min}: {stats_por_region.loc[region_min, "sum"]:.2f}')

# Calcular rango de ventas por región y producto (máximo - mínimo)
print('\nRango de ventas (max - min) por región y producto:')
def rango(x):
    return x.max() - x.min()

rango_ventas = df_ventas.groupby(['region', 'producto'])['ventas'].agg(['min', 'max', rango])
rango_ventas.columns = ['Mínimo', 'Máximo', 'Rango']
display(rango_ventas)

# Función de agregación personalizada adicional
print('\nCoeficiente de variación por región (desviación estándar / media):')
cv_por_region = df_ventas.groupby('region')['ventas'].agg(lambda x: x.std() / x.mean())
display(cv_por_region.sort_values(ascending=False))

# Agregaciones múltiples con funciones personalizadas
print('\nEstadísticas avanzadas por región:')
display(df_ventas.groupby('region')['ventas'].agg([
    ('Total', 'sum'),
    ('Promedio', 'mean'),
    ('Desv_Std', 'std'),
    ('Rango', rango),
    ('Conteo', 'count')
]))
````

Utiliza el archivo FoodMarket.xlsx que puede contener múltiples hojas para practicar merge:
- Carga las diferentes hojas del archivo FoodMarket.xlsx
- Identifica las relaciones entre las hojas (keys comunes)
- Realiza un merge interno (inner join) entre las hojas relacionadas
- Realiza un merge externo (outer join) para incluir todos los registros
- Realiza un merge por la izquierda (left join) usando una hoja como base
- Analiza las diferencias en los resultados obtenidos

Este ejercicio te ayudará a practicar merge con datos estructurados reales.

````python
import pandas as pd
# Ejercicio 8.2: Merge con datos de FoodMarket

food_path = '../../data/FoodMarket.xlsx'

# Cargar las diferentes hojas del archivo
xls = pd.ExcelFile(food_path)
print('Hojas disponibles en FoodMarket.xlsx:', xls.sheet_names)

# Cargar las hojas
hojas = {}
for hoja in xls.sheet_names:
    hojas[hoja] = pd.read_excel(food_path, sheet_name=hoja)
    print(f'\n--- Hoja: {hoja} ---')
    print(f'Shape: {hojas[hoja].shape}')
    print(f'Columnas: {list(hojas[hoja].columns)}')
    display(hojas[hoja].head(3))

# Identificar las claves comunes entre hojas
# Asumiendo que hay hojas relacionadas (por ejemplo: ventas, productos, clientes)
# Vamos a demostrar diferentes tipos de merge

# Si hay al menos 2 hojas, realizamos merge
if len(hojas) >= 2:
    hoja_names = list(hojas.keys())
    df1 = hojas[hoja_names[0]]
    df2 = hojas[hoja_names[1]] if len(hojas) > 1 else df1.copy()
    
    # Buscar columnas comunes
    columnas_comunes = set(df1.columns).intersection(set(df2.columns))
    print(f'\n\nColumnas comunes entre {hoja_names[0]} y {hoja_names[1]}: {columnas_comunes}')
    
    # Si hay columnas comunes, realizar merge
    if columnas_comunes:
        key = list(columnas_comunes)[0]
        
        # Merge interno (inner join)
        print(f'\n--- INNER JOIN (usando {key}) ---')
        merge_inner = pd.merge(df1, df2, on=key, how='inner')
        print(f'Shape resultado: {merge_inner.shape}')
        display(merge_inner.head())
        
        # Merge externo (outer join)
        print(f'\n--- OUTER JOIN (usando {key}) ---')
        merge_outer = pd.merge(df1, df2, on=key, how='outer')
        print(f'Shape resultado: {merge_outer.shape}')
        display(merge_outer.head())
        
        # Merge por la izquierda (left join)
        print(f'\n--- LEFT JOIN (usando {key}) ---')
        merge_left = pd.merge(df1, df2, on=key, how='left')
        print(f'Shape resultado: {merge_left.shape}')
        display(merge_left.head())
        
        # Análisis de diferencias
        print('\n--- ANÁLISIS DE DIFERENCIAS ---')
        print(f'Registros en df1: {len(df1)}')
        print(f'Registros en df2: {len(df2)}')
        print(f'Registros en inner join: {len(merge_inner)}')
        print(f'Registros en outer join: {len(merge_outer)}')
        print(f'Registros en left join: {len(merge_left)}')
    else:
        print('\nNo se encontraron columnas comunes para hacer merge automático.')
else:
    print('\nEl archivo solo tiene una hoja, creando DataFrames de ejemplo para merge.')
    # Crear DataFrames de ejemplo
    ventas = pd.DataFrame({
        'producto_id': [1, 2, 3, 4],
        'cantidad': [100, 150, 200, 120],
        'precio': [10.5, 15.0, 12.5, 8.0]
    })
    productos = pd.DataFrame({
        'producto_id': [1, 2, 3, 5],
        'nombre': ['Manzanas', 'Naranjas', 'Plátanos', 'Peras'],
        'categoria': ['Frutas', 'Frutas', 'Frutas', 'Frutas']
    })
    
    print('\nMerge inner:')
    display(pd.merge(ventas, productos, on='producto_id', how='inner'))
    print('\nMerge outer:')
    display(pd.merge(ventas, productos, on='producto_id', how='outer'))
    print('\nMerge left:')
    display(pd.merge(ventas, productos, on='producto_id', how='left'))
````

preferencias = {

````python
import pandas as pd
# Ejercicio 9.1: Tablas cruzadas básicas

preferencias = {
    'usuario': ['U001', 'U002', 'U003', 'U004', 'U005', 'U006', 'U007', 'U008'],
    'genero': ['F', 'M', 'F', 'M', 'F', 'M', 'F', 'M'],
    'edad': [25, 30, 35, 28, 32, 27, 33, 29],
    'producto_pref': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B']
}
df_pref = pd.DataFrame(preferencias)

print('DataFrame de preferencias:')
display(df_pref)

# Tabla cruzada de género vs preferencia de producto
print('\n--- TABLA CRUZADA: Género vs Producto Preferido ---')
tabla_genero_producto = pd.crosstab(df_pref['genero'], df_pref['producto_pref'])
display(tabla_genero_producto)

# Tabla cruzada de edad vs género (agrupando edades en rangos)
print('\n--- TABLA CRUZADA: Edad (rangos) vs Género ---')
# Crear rangos de edad
df_pref['rango_edad'] = pd.cut(df_pref['edad'], bins=[20, 27, 30, 35], labels=['20-27', '27-30', '30-35'])
tabla_edad_genero = pd.crosstab(df_pref['rango_edad'], df_pref['genero'])
display(tabla_edad_genero)

# Tabla cruzada con márgenes (totales)
print('\n--- TABLA CRUZADA CON MÁRGENES (totales) ---')
tabla_con_margenes = pd.crosstab(
    df_pref['genero'], 
    df_pref['producto_pref'], 
    margins=True, 
    margins_name='Total'
)
display(tabla_con_margenes)

# Tabla cruzada adicional: Género vs Producto con porcentajes
print('\n--- TABLA CRUZADA: Género vs Producto (porcentajes por fila) ---')
tabla_porcentajes = pd.crosstab(
    df_pref['genero'], 
    df_pref['producto_pref'], 
    normalize='index'
) * 100
display(tabla_porcentajes.round(2))

# Tabla cruzada con valores personalizados
print('\n--- TABLA CRUZADA: Conteo de usuarios por género y producto ---')
tabla_usuarios = pd.crosstab(
    df_pref['genero'], 
    df_pref['producto_pref'], 
    values=df_pref['edad'],
    aggfunc='count'
)
display(tabla_usuarios)
````

### Ejercicio 9.3: Análisis de frecuencias con grouping

Para el DataFrame de preferencias:

- Agrupa por producto preferido y calcula estadísticas de edad
- Crea grupos de edad (18-25, 26-30, 31-35) y analiza las preferencias por grupo

````python
import pandas as pd
# Ejercicio 9.3: Análisis de frecuencias con grouping

# Usar el DataFrame de preferencias
preferencias = {
    'usuario': ['U001', 'U002', 'U003', 'U004', 'U005', 'U006', 'U007', 'U008'],
    'genero': ['F', 'M', 'F', 'M', 'F', 'M', 'F', 'M'],
    'edad': [25, 30, 35, 28, 32, 27, 33, 29],
    'producto_pref': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B']
}
df_pref = pd.DataFrame(preferencias)

# Agrupa por género y cuenta las frecuencias
print('--- FRECUENCIAS POR GÉNERO ---')
freq_genero = df_pref['genero'].value_counts()
display(freq_genero)

# Alternativa con groupby
print('\nFrecuencias por género (usando groupby):')
freq_genero_groupby = df_pref.groupby('genero').size()
display(freq_genero_groupby)

# Porcentajes por género
print('\nPorcentajes por género:')
display((df_pref['genero'].value_counts(normalize=True) * 100).round(2))

# Agrupa por producto preferido y calcula estadísticas de edad
print('\n--- ESTADÍSTICAS DE EDAD POR PRODUCTO PREFERIDO ---')
stats_producto = df_pref.groupby('producto_pref')['edad'].agg([
    ('count', 'count'),
    ('edad_min', 'min'),
    ('edad_max', 'max'),
    ('edad_promedio', 'mean'),
    ('edad_mediana', 'median'),
    ('desv_std', 'std')
])
display(stats_producto.round(2))

# Frecuencias de producto preferido
print('\nFrecuencias de producto preferido:')
display(df_pref['producto_pref'].value_counts())

# Crea grupos de edad (18-25, 26-30, 31-35) y analiza las preferencias por grupo
print('\n--- ANÁLISIS DE PREFERENCIAS POR GRUPO DE EDAD ---')
df_pref['grupo_edad'] = pd.cut(
    df_pref['edad'], 
    bins=[18, 25, 30, 35], 
    labels=['18-25', '26-30', '31-35'],
    include_lowest=True
)

print('DataFrame con grupos de edad:')
display(df_pref)

# Frecuencias por grupo de edad
print('\nFrecuencias por grupo de edad:')
display(df_pref['grupo_edad'].value_counts().sort_index())

# Preferencias por grupo de edad
print('\nPreferencias de producto por grupo de edad:')
tabla_grupo_edad = pd.crosstab(df_pref['grupo_edad'], df_pref['producto_pref'])
display(tabla_grupo_edad)

# Tabla cruzada con porcentajes por fila
print('\nPreferencias de producto por grupo de edad (porcentajes):')
tabla_grupo_edad_pct = pd.crosstab(df_pref['grupo_edad'], df_pref['producto_pref'], normalize='index') * 100
display(tabla_grupo_edad_pct.round(2))

# Análisis combinado: género, grupo de edad y producto
print('\n--- ANÁLISIS COMBINADO: Género y Grupo de Edad ---')
tabla_combinada = df_pref.groupby(['grupo_edad', 'genero', 'producto_pref']).size().unstack(fill_value=0)
display(tabla_combinada)

# Frecuencias con múltiples columnas
print('\nFrecuencias por género y producto:')
freq_multi = df_pref.groupby(['genero', 'producto_pref']).size()
display(freq_multi)
````

Dado un DataFrame en formato ancho (ventas mensuales por producto), conviértelo a formato largo:

ventas_ancho = {

````python
import pandas as pd
# Ejercicio 10.1: Convertir de formato ancho a largo

ventas_ancho = {
    'producto': ['Producto_A', 'Producto_B', 'Producto_C'],
    'enero': [100, 150, 200],
    'febrero': [120, 180, 190],
    'marzo': [110, 160, 210],
    'abril': [130, 170, 220]
}
df_ancho = pd.DataFrame(ventas_ancho)

print('DataFrame en formato ANCHO:')
display(df_ancho)

# Convertir a formato largo usando melt
# 'producto' se mantiene fija, los meses se convierten en filas
print('\n--- CONVERSIÓN A FORMATO LARGO (melt) ---')
df_largo = pd.melt(
    df_ancho,
    id_vars=['producto'],
    var_name='mes',
    value_name='ventas'
)
display(df_largo)

# Ordenar por producto para mejor visualización
print('\nDataFrame largo ordenado por producto:')
df_largo_ordenado = df_largo.sort_values(['producto', 'mes']).reset_index(drop=True)
display(df_largo_ordenado)

# Alternativa: especificando explícitamente las columnas a transformar
print('\n--- MELT especificando columnas de valores ---')
df_largo_alt = pd.melt(
    df_ancho,
    id_vars=['producto'],
    value_vars=['enero', 'febrero', 'marzo', 'abril'],
    var_name='mes',
    value_name='cantidad_vendida'
)
display(df_largo_alt.head(10))

# Información sobre la transformación
print('\n--- INFORMACIÓN DE LA TRANSFORMACIÓN ---')
print(f'Formato ancho: {df_ancho.shape[0]} filas x {df_ancho.shape[1]} columnas')
print(f'Formato largo: {df_largo.shape[0]} filas x {df_largo.shape[1]} columnas')
print(f'Columnas fijas: producto')
print(f'Columnas transformadas: enero, febrero, marzo, abril')
print(f'Total de registros en formato largo: {len(df_largo)}')
````

### Ejercicio 10.3: Reestructuración avanzada

Crea un DataFrame con múltiples variables que necesite reestructuración compleja:
- Incluye tanto datos cuantitativos como cualitativos
- Aplica melt con parámetros específicos
- Aplica pivot_table para agregar datos
- Compara los resultados obtenidos

````python
import pandas as pd
import numpy as np
# Ejercicio 10.3: Reestructuración avanzada

# Crear un DataFrame complejo con múltiples variables
datos_ventas = {
    'tienda': ['T1', 'T1', 'T2', 'T2', 'T3', 'T3'] * 2,
    'producto': ['A', 'B', 'A', 'B', 'A', 'B'] * 2,
    'trimestre': ['Q1']*6 + ['Q2']*6,
    'ventas': [100, 150, 120, 180, 90, 160, 110, 170, 130, 190, 95, 175],
    'costos': [60, 90, 70, 110, 55, 95, 65, 100, 75, 115, 58, 105],
    'categoria': ['Electrónica', 'Electrónica', 'Electrónica', 'Electrónica', 'Electrónica', 'Electrónica'] * 2
}
df_complejo = pd.DataFrame(datos_ventas)

print('DataFrame complejo original:')
display(df_complejo)

# Aplicar melt con parámetros específicos
print('\n--- MELT CON PARÁMETROS ESPECÍFICOS ---')
# Mantener fijas: tienda, producto, trimestre, categoria
# Convertir a formato largo: ventas y costos
df_melt = pd.melt(
    df_complejo,
    id_vars=['tienda', 'producto', 'trimestre', 'categoria'],
    value_vars=['ventas', 'costos'],
    var_name='metrica',
    value_name='valor'
)
print('DataFrame después de melt:')
display(df_melt.head(15))

# Aplicar pivot_table para agregar datos
print('\n--- PIVOT_TABLE: Ventas promedio por tienda y producto ---')
pivot_ventas = pd.pivot_table(
    df_complejo,
    values='ventas',
    index='tienda',
    columns='producto',
    aggfunc='mean'
)
display(pivot_ventas)

# Pivot table con múltiples valores
print('\n--- PIVOT_TABLE: Múltiples métricas (ventas y costos) ---')
pivot_multi = pd.pivot_table(
    df_complejo,
    values=['ventas', 'costos'],
    index='tienda',
    columns='producto',
    aggfunc='sum'
)
display(pivot_multi)

# Pivot table con márgenes (totales)
print('\n--- PIVOT_TABLE con MÁRGENES (totales) ---')
pivot_margenes = pd.pivot_table(
    df_complejo,
    values='ventas',
    index='tienda',
    columns='producto',
    aggfunc='sum',
    margins=True,
    margins_name='Total'
)
display(pivot_margenes)

# Pivot table con múltiples funciones de agregación
print('\n--- PIVOT_TABLE: Múltiples funciones de agregación ---')
pivot_multiagg = pd.pivot_table(
    df_complejo,
    values='ventas',
    index='tienda',
    columns='producto',
    aggfunc=['sum', 'mean', 'max', 'min']
)
display(pivot_multiagg)

# Comparación de resultados
print('\n--- COMPARACIÓN DE RESULTADOS ---')
print(f'DataFrame original: {df_complejo.shape}')
print(f'DataFrame después de melt: {df_melt.shape}')
print(f'Pivot table (ventas promedio): {pivot_ventas.shape}')
print(f'Pivot table (multi valores): {pivot_multi.shape}')

# Reestructuración compleja: de melt a pivot
print('\n--- REESTRUCTURACIÓN COMPLEJA: Melt → Pivot ---')
# Primero, crear formato largo
df_largo_complejo = pd.melt(
    df_complejo,
    id_vars=['tienda', 'producto', 'trimestre'],
    value_vars=['ventas', 'costos'],
    var_name='tipo',
    value_name='cantidad'
)

# Luego, pivotar para tener ventas y costos como columnas
df_pivot_complejo = df_largo_complejo.pivot_table(
    values='cantidad',
    index=['tienda', 'producto'],
    columns='tipo',
    aggfunc='sum'
)
display(df_pivot_complejo)

# Calcular margen de ganancia
print('\nCálculo de margen (ventas - costos):')
df_pivot_complejo['margen'] = df_pivot_complejo['ventas'] - df_pivot_complejo['costos']
df_pivot_complejo['porcentaje_margen'] = (df_pivot_complejo['margen'] / df_pivot_complejo['ventas'] * 100).round(2)
display(df_pivot_complejo)
````
