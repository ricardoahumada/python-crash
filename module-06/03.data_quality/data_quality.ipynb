# Análisis y Limpieza de Datos

## Calidad de Datos

La **calidad de datos** se refiere al grado en que los datos son aptos para su uso previsto en operaciones, toma de decisiones y planificación. Datos de alta calidad son precisos, completos, consistentes y relevantes para el contexto en el que se utilizan.

### Características de Datos de Calidad

- **Exactitud**: Los datos reflejan correctamente la realidad que representan
- **Completitud**: No existen valores faltantes o nulos en campos críticos
- **Consistencia**: Los datos son coherentes entre diferentes fuentes y a lo largo del tiempo
- **Validez**: Los datos cumplen con las reglas de negocio y restricciones definidas
- **Actualidad**: Los datos están actualizados y son relevantes para el momento actual
- **Unicidad**: No existen registros duplicados que distorsionen el análisis

### Tipos de No Calidades (Problemas Comunes)

1. **Valores faltantes (Missing values)**: Campos vacíos o nulos que impiden análisis completos
2. **Valores atípicos (Outliers)**: Datos extremos que pueden ser errores de captura o casos excepcionales
3. **Duplicados**: Registros repetidos que sesgan estadísticas y conteos
4. **Inconsistencias**: Datos contradictorios (ej. fecha de envío anterior a fecha de pedido)
5. **Formatos incorrectos**: Tipos de datos inadecuados (texto en lugar de número, fechas mal formateadas)
6. **Datos desactualizados**: Información obsoleta que no refleja la situación actual
7. **Valores fuera de rango**: Datos que no cumplen con restricciones lógicas del negocio

## Implicaciones de las No Calidades en Procesados Posteriores

### 1. **Valores faltantes (Missing values)**
**Implicaciones:**
- **Modelos de Machine Learning**: Muchos algoritmos no aceptan valores nulos y generarán errores
- **Cálculos estadísticos**: Promedios, sumas y agregaciones producirán resultados sesgados o incorrectos
- **Visualizaciones**: Gráficos incompletos o distorsionados que pueden llevar a conclusiones erróneas
- **Joins y cruces**: Pérdida de registros al combinar datasets si los valores nulos están en claves de unión
- **Segmentación**: Imposibilidad de clasificar o agrupar registros adecuadamente

### 2. **Valores atípicos (Outliers)**
**Implicaciones:**
- **Modelos predictivos**: Distorsión en regresiones lineales, árboles de decisión y otros algoritmos sensibles
- **Estadísticas descriptivas**: Media y desviación estándar sesgadas, dando una visión falsa de la distribución
- **Detección de patrones**: Enmascaramiento de tendencias reales por valores extremos
- **Escalado de datos**: Normalización y estandarización afectadas, reduciendo efectividad de modelos
- **Presupuestos y forecasting**: Proyecciones financieras incorrectas basadas en datos extremos no representativos

### 3. **Duplicados**
**Implicaciones:**
- **Conteos y métricas**: Inflación artificial de KPIs (ventas totales, número de clientes, etc.)
- **Análisis de frecuencia**: Sesgos en distribuciones y rankings de productos/clientes más vendidos
- **Modelos de ML**: Sobreajuste (overfitting) al dar mayor peso a ciertos patrones repetidos
- **Segmentación de clientes**: Clientes contados múltiples veces distorsionan perfiles y campañas de marketing
- **Costos operativos**: Envíos duplicados, comunicaciones redundantes a los mismos clientes

### 4. **Inconsistencias**
**Implicaciones:**
- **Cálculos de métricas de tiempo**: KPIs como "tiempo de entrega" con valores negativos o imposibles
- **Validaciones de negocio**: Fallo en reglas de integridad referencial y lógica de procesos
- **Confianza en los datos**: Pérdida de credibilidad en reportes y dashboards gerenciales
- **Auditorías**: Problemas de cumplimiento normativo y trazabilidad de operaciones
- **Automatizaciones**: Procesos ETL y pipelines que fallan o producen resultados erróneos

### 5. **Formatos incorrectos**
**Implicaciones:**
- **Errores de procesamiento**: Scripts y consultas SQL que fallan al intentar operaciones numéricas en texto
- **Ordenamiento incorrecto**: Fechas como texto se ordenan alfabéticamente en lugar de cronológicamente
- **Incompatibilidad entre sistemas**: Problemas al integrar datos de diferentes fuentes con formatos distintos
- **Pérdida de precisión**: Números almacenados como texto pierden capacidad de cálculo matemático
- **Consumo de recursos**: Mayor uso de memoria y procesamiento al convertir tipos constantemente

### 6. **Datos desactualizados**
**Implicaciones:**
- **Decisiones erróneas**: Estrategias de negocio basadas en información que ya no refleja la realidad
- **Análisis de tendencias**: Imposibilidad de detectar cambios recientes en comportamiento de clientes
- **Modelos obsoletos**: Algoritmos entrenados con datos antiguos tienen bajo rendimiento predictivo
- **Satisfacción del cliente**: Envíos a direcciones antiguas, contactos a números inactivos
- **Oportunidades perdidas**: Incapacidad de reaccionar a tiempo ante cambios del mercado

### 7. **Valores fuera de rango**
**Implicaciones:**
- **Validación de modelos**: Predicciones imposibles (ej. edad de -5 años, cantidad negativa de productos)
- **Alertas y monitoreo**: Sistemas de detección de anomalías generan falsos positivos o negativos
- **Integridad de reportes**: Métricas clave (márgenes, porcentajes) con valores ilógicos que invalidan análisis
- **Problemas en aplicaciones**: Errores en interfaces de usuario que esperan rangos válidos
- **Cumplimiento normativo**: Violación de restricciones legales (ej. descuentos superiores al 100%)

## Proceso de Limpieza de Datos

Este proceso de limpieza aborda sistemáticamente los principales problemas de calidad identificados en el dataset de ventas:

### 1. **Eliminación de Columnas No Significativas**
- Por alta proporción de valores nulos y baja relevancia analítica
- Reducción de dimensionalidad sin pérdida de información crítica

### 2. **Tratamiento de Valores Faltantes**
- **Eliminación**: Filas con valores nulos en campos críticos
- **Imputación numérica**: media o mediana para columnas numéricas.
- **Imputación categórica**: Moda para coluymnas categóricas.

### 3. **Detección y Tratamiento de Outliers**
- Análisis mediante método IQR (Rango Intercuartílico) para variables numéricas clave
- Eliminación de valores extremos
- Preservación de datos atípicos legítimos dentro de rangos de negocio razonables

### 4. **Corrección de Tipos de Datos**
- Conversión de columnas de fechas a formato datetime
- Permite operaciones temporales precisas y cálculos de diferencias de días

### 5. **Eliminación de Duplicados**
- Identificación y remoción de registros completamente duplicados
- Garantiza unicidad en transacciones y evita inflación de métricas

### 6. **Validación de Consistencia**
- Verificación de relaciones lógicas entre fechas (envío posterior a pedido)
- Comprobación de categorías válidas

### 7. **Enriquecimiento de Datos**
- Creación de variables derivadas para enriquecer análisis
- Métricas de tiempo útiles para análisis de eficiencia operativa

### 8. **Exportación y Persistencia**
- Dataset limpio guardado en formato CSV para análisis posteriores
- Reset de índices para mantener integridad estructural

**Resultado:** Dataset limpio, consistente y listo para modelado predictivo, análisis estadístico y visualizaciones confiables.

```python
# Identificar y eliminar columnas no significativas
# Eliminar la columna 'comments' ya que tiene muchos valores nulos y no es crítica para el análisis
df = df.drop(columns=['comments'])

print("Columnas eliminadas: ['comments']")
print(f"Dimensiones del dataframe: {df.shape}")
```

```python
# # Mostrar las primeras filas
# display(df.head())

# # Información general del dataset
# display(df.info())

# Estadísticas descriptivas
# display(df.describe(include='all'))
display(df.describe(include=np.number))
```

```python
display(df.describe(include=['object']))
```

```python
# Contar valores nulos por columna
print("Valores nulos por columna:")
print(df.isnull().sum())
```

```python
# Identificar columnas con valores nulos y decidir estrategia
print("Porcentaje de valores nulos por columna:")
null_percentage = (df.isnull().sum() / len(df)) * 100
print(null_percentage[null_percentage > 0].sort_values(ascending=False))

# Eliminar filas donde faltan datos críticos (ordernumber, orderdate, productcode)
df = df.dropna(subset=['ordernumber', 'orderdate', 'productcode'])

# Rellenar valores numéricos con la mediana
df['customernumber'] = df['customernumber'].fillna(df['customernumber'].median())
df['employeenumber'] = df['employeenumber'].fillna(df['employeenumber'].median())
df['orderlinenumber'] = df['orderlinenumber'].fillna(df['orderlinenumber'].median())

# Rellenar status con el valor más frecuente
df['status'] = df['status'].fillna(df['status'].mode()[0])

print(f"\nFilas restantes después de limpieza: {len(df)}")
print("\nValores nulos restantes:")
print(df.isnull().sum())
```

```python
# Verificar outliers en columnas numéricas clave
numeric_cols = ['quantityordered', 'priceeach', 'sales_amount']

for col in numeric_cols:
    print(f"\n--- Análisis de outliers en '{col}' ---")
    print(f"Min: {df[col].min()}")
    print(f"Max: {df[col].max()}")
    print(f"Media: {df[col].mean():.2f}")
    print(f"Mediana: {df[col].median():.2f}")
    
    # Método IQR
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers_iqr = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"Outliers (IQR): {len(outliers_iqr)} valores fuera de [{lower_bound:.2f}, {upper_bound:.2f}]")
```

```python
# Eliminar outliers extremos (ajustar según contexto)
df = df[df['priceeach'] <= 10000]
df = df[df['quantityordered'] <= 1000]
```

```python
# Verificar tipos de datos
print("Tipos de datos actuales:")
print(df.dtypes)

# Convertir fechas a tipo datetime si no lo están
date_cols = ['orderdate', 'shippeddate', 'requireddate']
for col in date_cols:
    if df[col].dtype != 'datetime64[ns]':
        df[col] = pd.to_datetime(df[col], errors='coerce')

df.info()
```

```python
# Buscar duplicados exactos
duplicates = df[df.duplicated()]
print("\nNúmero de filas duplicadas:", len(duplicates))

# Eliminar duplicados (opcional, revisar antes)
df.drop_duplicates(inplace=True)
```

```python
# Verificar categorías únicas en columnas relevantes
print("Valores únicos en 'status':", df['status'].unique())
print("Valores únicos en 'productCode':", df['productcode'].nunique())
```

```python
# Comprobamos que shippedDate no sea antes de orderDate
invalid_shipped = df[(df['shippeddate'] < df['orderdate'])
                     & df['shippeddate'].notnull()]
print("Pedidos con fecha de envío inválida:", len(invalid_shipped))
```

```python
# Rellenar valores faltantes en sales_amount usando quantityOrdered y priceEach
df['sales_amount'] = df['sales_amount'].fillna(df['quantityordered'] * df['priceeach'])

# df['real_cost_amount'] = df['quantityordered'] * df['priceeach']
```

```python
# Calcular días entre orden y envío (si ya fue enviado)
df['days_to_ship'] = (df['shippeddate'] - df['orderdate']).dt.days

# Calcular días entre orden y fecha requerida
df['days_required'] = (df['requireddate'] - df['orderdate']).dt.days
```

```python
display(df)
```

```python
# Resetear índice (opcional)
df.reset_index(drop=True, inplace=True)

# Exportar dataset limpio
df.to_csv("../output/ventas_limpias.csv", index=False)
print("Datos limpios guardados en 'ventas_limpias.csv'")
```
