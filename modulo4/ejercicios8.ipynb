```python
class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura
    
    def area(self):
        return self.base * self.altura
    
    def perimetro(self):
        return 2 * (self.base + self.altura)
    
    def es_cuadrado(self):
        return self.base == self.altura
```

```python
r1 = Rectangulo(5, 4)
print(f"Rectángulo 5x4 - Área: {r1.area()}, Perímetro: {r1.perimetro()}, ¿Es cuadrado?: {r1.es_cuadrado()}")

r2 = Rectangulo(5, 5)
print(f"Rectángulo 5x5 - Área: {r2.area()}, Perímetro: {r2.perimetro()}, ¿Es cuadrado?: {r2.es_cuadrado()}")
```

```python
# Ejercicio 2: Herencia
class Vehiculo:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo
    
    def descripcion(self):
        return f"{self.marca} {self.modelo}"
```

```python
class Auto(Vehiculo):
    def __init__(self, marca, modelo, puertas):
        super().__init__(marca, modelo)
        self.puertas = puertas
    
    def descripcion(self):
        return f"{super().descripcion()} - {self.puertas} puertas"
```

```python
class Moto(Vehiculo):
    def __init__(self, marca, modelo, cilindrada):
        super().__init__(marca, modelo)
        self.cilindrada = cilindrada
    
    def descripcion(self):
        return f"{super().descripcion()} - {self.cilindrada}cc"
```

```python
auto = Auto("Toyota", "Corolla", 4)
moto = Moto("Yamaha", "MT-07", 689)

print(auto.descripcion())
print(moto.descripcion())
```

```python
# Ejercicio 3: Composición
class Direccion:
    def __init__(self, calle, ciudad, pais):
        self.calle = calle
        self.ciudad = ciudad
        self.pais = pais

    def __str__(self):
        return f"{self.calle}, {self.ciudad}, {self.pais}"
```

```python
class Persona:
    __direccion = None

    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def agregar_direccion(self, direccion):
        self.__direccion = direccion

    def mostrar_direccion_completa(self):
        if self.__direccion:
            return f"{self.__direccion}"
        else:
            return "No se ha agregado una dirección"
```

```python
direccion = Direccion("Av. Principal 123", "Buenos Aires", "Argentina")
persona = Persona("Carlos", 30)

print(persona.mostrar_direccion_completa())

persona.agregar_direccion(direccion)
print(persona.mostrar_direccion_completa())
```
