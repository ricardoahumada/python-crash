## Comprehensions

```python
m = []
for n in range(5):
    m.append(n**2)

m
```

```python
m = [n**2 for n in range(5)]
m
```

```python
m = []
for n in range(5):
    if n % 2 == 0:
        m.append(n**2)

m
```

```python
m = [n**2 for n in range(5) if (n % 2 == 0)]
m
```

```python
cities = ["Madrid", "Barcelona", "Milán", "Santander"]
years = [2017, 2018, 2019, 2020]
```

```python
a = []
for c in cities:
    for y in years:
        a.append((c, y))
a
```

```python
[(c, y) for c in cities for y in years]
```

```python
[(a*2, b, c, d) for a in ["a1", "a2"] for b in ["b1", "b2"] for c in ["c1", "c2"] for d in ["d1", "d2"]]
```

```python
cities = ["Madrid", "Barcelona", "Milán", "Santander"]
years = [2017, 2018, 2019, 2020]

[(c, y) for c in cities for y in years if c.startswith("M") if y % 2 == 0]
```

```python
{(c, y) for c in cities for y in years if c.startswith("M") if y % 2 == 0}
```

```python
cities = ["Madrid", "Barcelona", "Milán", "Santander"]
{c: len(c) for c in cities}
```

```python
cities = ["Madrid", "Barcelona", "Milán", "Santander"]
years = [2017, 2018, 2019, 2020]

{c : y for c in cities for y in years if y % 2 == 0}
```
