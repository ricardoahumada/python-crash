# Ejercicio Integrador — Gestor de Finanzas Personales


## Contexto

Una persona quiere llevar el control de sus ingresos y gastos mensuales desde el terminal, sin depender de apps externas. Hoy lo hace en una hoja de cálculo y se le desordena: pierde el histórico, no puede filtrar por categoría, no sabe en qué se le va la plata.

Tu equipo tiene que construir una herramienta que cubra ese problema. **Ustedes deciden** cómo modelar los datos, cómo se almacenan y qué operaciones ofrecer. Lo que sigue son los **requerimientos** y los **criterios de aceptación** que tu solución tiene que cumplir.

---

## Requerimientos

### 1. Modelo de datos

Cada movimiento registrado debe tener, como mínimo, esta información:

- Un **identificador único** que se asigne automáticamente y no se reutilice dentro de la misma sesión.
- Su **tipo**: solo se admiten `ingreso` y `gasto`. Cualquier otro valor debe rechazarse.
- Un **monto** numérico positivo, en euros. Montos ≤ 0 deben rechazarse.
- Una **categoría** elegida de un conjunto acotado de categorías válidas que ustedes definan (mínimo 5, separadas por dominio: por ejemplo alimentación, transporte, ocio, etc.). Categorías fuera del conjunto deben rechazarse.
- Una **fecha** válida (sugerido formato `YYYY-MM-DD`, pero pueden elegir otro siempre que sea consistente y validable).
- Una **descripción** textual no vacía.

### 2. Encapsulamiento

Toda la lógica de negocio —registro, consultas, estadísticas, persistencia— debe vivir dentro de **una sola clase**. No se acepta un script de funciones sueltas, ni un `main()` con la lógica inlineada.

El `main()` del programa debe limitarse a mostrar el menú, leer la opción del usuario y delegar en la clase.

### 3. Operaciones mínimas

La clase debe permitir, como mínimo:

- **Registrar** un movimiento nuevo con todos los campos validados.
- **Listar / filtrar** movimientos. Al menos dos de los siguientes criterios de filtro tienen que estar disponibles: por tipo, por categoría, por rango de fechas.
- **Calcular estadísticas** sobre los datos. Al menos:
  - Balance general (ingresos − gastos).
  - Total por tipo y/o por categoría.
  - Un ranking de las categorías con mayor gasto, ordenable.
- **Persistir** los datos en disco, en el formato que elijan, de modo que al cerrar y volver a abrir el programa la información se recupere.
- **Manejar el caso de primera ejecución** (archivo inexistente) sin crashear.

### 4. Interfaz

La interacción con el usuario se hace por **terminal**, con un menú de opciones numeradas y un bucle que pida la opción hasta que el usuario elija "salir". La salida debe ser legible y formateada con f-strings o `.format()` (queda a elección del equipo).

### 5. Calidad

- El código debe seguir **PEP 8** de forma razonable: nombres en `snake_case`, funciones y clases documentadas con docstring, líneas cortas, indentación consistente.
- **Solo se permite la biblioteca estándar** de Python (no `pandas`, `numpy`, ni librerías externas).
- Las decisiones de diseño que no estén explícitamente fijadas arriba (formato de archivo, nombres de métodos, layout del menú, qué hacer con los IDs al recargar desde disco, etc.) las toma el equipo. **Justifiquen en un comentario al inicio del archivo** las decisiones no triviales que tomaron.

---

## Criterios de aceptación

Tu solución se considera completa cuando cumple **todos** estos puntos verificables:

- [ ] El archivo se ejecuta con `python gestor_finanzas.py` y abre un menú interactivo.
- [ ] Toda la lógica de negocio está en una sola clase.
- [ ] Intentar registrar un movimiento con tipo, monto, categoría, fecha o descripción inválidos produce un error claro y controlado (no un `Traceback` críptico).
- [ ] Al menos dos filtros distintos funcionan y devuelven los resultados esperados.
- [ ] Las estadísticas básicas se calculan correctamente y se pueden verificar con datos conocidos.
- [ ] Los datos sobreviven al cierre del programa: registrar movimientos, salir, volver a entrar y comprobar que siguen ahí.
- [ ] La primera vez que se ejecuta (sin archivo previo) no rompe ni muestra errores.
- [ ] PEP 8 razonable y docstrings presentes en la clase y sus métodos públicos.
- [ ] Solo se usa la biblioteca estándar.

## Bonus (opcional, solo si sobra tiempo)

- Exportar un reporte a **CSV** con `csv` de la stdlib.
- Un **gráfico de barras ASCII** del gasto por categoría, renderizado en la terminal.
- Cualquier extensión que el equipo considere útil (presupuesto mensual por categoría, alertas por sobre-categoría, búsqueda por texto en descripción, etc.) siempre que se sostenga con la misma calidad que el resto.
