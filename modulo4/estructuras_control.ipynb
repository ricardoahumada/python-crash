# IF

`````python
n = 2
m = 2
if n > m:
    print(n, "Es mayor que ", m)
    print("este es otro print...")
elif n == m:
    print(n, "Es igual que ", m)
else:
    print(n, "Es menor que ", m)
    print("este es otro print 2...")

print("fin de proceso")
`````

`````python
lista = [1,2,3,4,5]

for n in lista:
    print("valor:", n)
`````

`````python
lista_tuplas = [("a",1),("b",2),("c",3)]

for (k, v) in lista_tuplas:
    print(k,v)
`````

## While

`````python
m = 1

while m < 10:
    print(m)
    m += 1
    if m == 4:
        break
`````
