# POO

```python
class Cliente:
    # Atributos de clase
    # nombre = ''
    # direccion = ''
    # dni = ''
    # gasto = 0.0
    activo = False
    __descuento = 5.0  # cuando gasto mayor de 60€
    __IVA = 21.0

    # Constructor
    def __init__(self, nombre, direccion, dni, gasto, activo):
        self.nombre = nombre
        self.direccion = direccion
        self.dni = dni
        self.gasto = gasto
        self.activo = activo

    def __str__(self):
        return f"Cliente(nombre={self.nombre}, direccion={self.direccion}, dni={self.dni}, gasto={self.gasto}, activo={self.activo})"


    def __del__(self):
        print(f"El cliente {self.nombre} ha sido eliminado.")


    def __add__(self, other):
        if isinstance(other, Cliente):
            return self.gasto + other.gasto
        return NotImplemented


    # Métodos de la clase (Comportamientos)
    def darse_de_alta(self):
        self.activo = True
        self.gasto = 0.0

    def darse_de_baja(self):
        self.activo = False

    def realizar_gasto(self, importe):
        if self.activo:
            self.gasto += importe
        else:
            print(f"El cliente {self.nombre} no está activo. No se puede realizar la compra.")


    def __calcular_descuento(self, importe):
        total_con_descuento = importe

        if importe > 60:
            total_con_descuento = importe * (1 - self.__descuento / 100)

        return total_con_descuento


    def facturar(self):
        total_con_descuento = self.__calcular_descuento(self.gasto)
        totala_con_IVA = total_con_descuento * (1 + self.__IVA / 100)
        print(f"Total a pagar con IVA: {totala_con_IVA:.2f}€")
```

## Herencia / Subclases -> Clases especializadas

```python
netmind = Empresa("Netmind", "Calle Netmind 789", "NIF12345678", 1000.0, True, "CIF87654321")
print(netmind)

netmind.realizar_gasto(100.0)

print(netmind)

netmind.facturar()
```

## Servicios

```python
class Facturadora:

    def __init__(self, lista_clientes: [Cliente]):
        self.clientes = lista_clientes

    def facturar_todos(self):
        if self.clientes:
            for cliente in self.clientes:
                print(f"Facturando cliente: {cliente.nombre}")
                cliente.facturar()
        else:
            print("No hay clientes para facturar.")
```

```python
f_clientes = Facturadora([anna, pedro])
print(f_clientes)

f_empresas = Facturadora([netmind])
print(f_empresas)
```

```python
f_clientes.facturar_todos()
```
