```python

def proceso_lista(una_lista):
    una_lista_interna = list(una_lista)
    # una_lista.append(8) antipatron
    una_lista_interna.append(8)
    return una_lista_interna
```

```python
lista = [1, 2, 3]
print(lista)

lista = proceso_lista(lista)

print(lista)
```

```python
def sumar(a, b):
    return a + b


def restar(a, b):
    return a - b


def sumar_restar(a, b):
    return sumar(a, b), restar(a, b)
```

```python
def sumar_restar(a, b):
    def _sumar(a, b):
        return a + b

    def _restar(a, b):
        return a - b

    return _sumar(a, b), _restar(a, b)
```

```python
sumar_restar(4,5)
```
