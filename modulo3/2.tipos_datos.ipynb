## Conjuntos

```python
un_conjunto = {1, 2, 3, 5, 4, '4', (1, 2), (1, 2)}
print(un_conjunto)
print(type(un_conjunto))
print(len(un_conjunto))
```

```python
print((1, 2) in un_conjunto)
```

```python
un_conjunto.add(6)
print(un_conjunto)
una_lista = [11, 12, 13, 13, 12, 11]
un_conjunto.update(una_lista)
print(un_conjunto)
```

```python
# un_conjunto.remove(2)
un_conjunto.discard(2)
print(un_conjunto)
extraido = un_conjunto.pop()
print(extraido)
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

c = a.union(b)
print(c)

d = a.intersection(b)
print(d)

d = a.difference(b)
print(d)

d = b.difference(a)
print(d)

d = b.symmetric_difference(a)
print(d)

print(a.issubset(c))
print(a.issubset(d))
print(c.issuperset(a))
```

```python
meses = ["ene","feb","mar","abr"]
facturas = [100, 200, 150, 300]
totales = (34565, 45678, 56789, 67890)

zipped = list(zip(meses, facturas, totales))

print(zipped)
print(dict(zip(meses, facturas)))
```
