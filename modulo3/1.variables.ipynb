# Operaciones con variables

```python
m = 3+2
n = 2
print(m, n)
```

```python
# Operaciones algebricas con variables
print(m+n)
print(m-n)
print(m*n)
print(m/n)
print(m//n)
print(m%n)
print(m**n)
```

```python
# Operaciones comparacion
print(m > n)
print(m < n)
print(m >= n)
print(m <= n)
print(m == n)
print(m != n)
```

```python
# Operaciones logicos
print((m > n) and (m < 10))
print(m > n or m < 10)
print(not (m > n))
```

```python
# Operaciones de petencia
texto = "Esto es un texto en Python"
print("texto" in texto)
print("Texto" in texto)
```

```python
s = set([1, 2, 2, 3])
print(s)
print(2 in s)
print(4 in s)
```

```python
# Operadores de indentidad
a = "texto"
b = "texto"

print(a == b)
print(a is b)
```

```python
n = [1, 2, 3]
m = [1, 2, 3]

print(n == m)
print(n is m)
```

```python
n = m
print(n == m)
print(n is m)
```

```python
# Operaciones a nivel de bit
b1 = 4
b2 = 5

print(bin(b1))
print(bin(b2))

print(b1 & b2)
print(b1 | b2)
```
