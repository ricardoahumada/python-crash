# Módulo 8: Automatización de Procesos
## Proyecto Integrador: Sistema de Reportes Automatizado

En este proyecto final integrador, combinaremos todas las técnicas aprendidas.

## Especificación del Proyecto

El sistema debe:

1. **Generar datos simulados** de ventas/indicadores
2. **Crear un informe HTML** con formato profesional
3. **Enviar por correo** (simulado)
4. **Comprimir y archivar** el reporte

## Paso 1: Estructura del Proyecto

```python
import os
carpetas = ['proyecto/datos', 'proyecto/informes', 'proyecto/respaldos']
for c in carpetas:
    os.makedirs(c, exist_ok=True)
    print(f"✓ {c}")
print("\nEstructura lista")
```

## Paso 2: Generar Datos Simulados

```python
import random
from datetime import datetime, timedelta

def generar_datos(dias=7):
    productos = ['A', 'B', 'C', 'D', 'E']
    precios = {'A': 25.50, 'B': 42.00, 'C': 15.75, 'D': 89.00, 'E': 35.20}
    datos = []
    hoy = datetime.now()
    
    for dia in range(dias):
        fecha = (hoy - timedelta(days=dia)).strftime('%Y-%m-%d')
        for prod in productos:
            datos.append({
                'fecha': fecha,
                'producto': f'Producto {prod}',
                'unidades': random.randint(50, 200),
                'precio': precios[prod],
                'total': random.randint(50, 200) * precios[prod]
            })
    return datos

datos = generar_datos()
print(f"Generados {len(datos)} registros")
print(f"Ejemplo: {datos[0]}")
```

## Paso 3: Generar Informe HTML

```python
from datetime import datetime

def generar_informe(datos, titulo):
    total = sum(d['total'] for d in datos)
    unidades = sum(d['unidades'] for d in datos)
    prods = len(set(d['producto'] for d in datos))
    
    filas = ""
    for d in datos:
        filas += f"<tr><td>{d['fecha']}</td><td>{d['producto']}</td><td style='text-align:right'>{d['unidades']}</td><td style='text-align:right'>€{d['total']:.2f}</td></tr>"
    
    html = f"""<!DOCTYPE html><html><head><style>
    body{font-family:'Segoe UI',Arial;margin:40px;background:#f5f6fa}
    .container{background:white;padding:30px;border-radius:10px}
    h1{color:#2c3e50;border-bottom:3px solid #3498db}
    .summary{display:flex;gap:20px;margin:20px 0}
    .card{flex:1;background:#f8f9fa;padding:20px;border-radius:8px;text-align:center;border-left:4px solid #3498db}
    .card h3{font-size:2em;color:#2c3e50;margin:0}
    table{width:100%;border-collapse:collapse;margin-top:20px}
    th{background:#2c3e50;color:white;padding:15px;text-align:left}
    td{padding:12px 15px;border-bottom:1px solid #ecf0f1}
    tr:hover{background:#f8f9fa}
</style></head><body>
<div class='container'>
    <h1>{titulo}</h1>
    <p>Fecha: {datetime.now().strftime('%d/%m/%Y')}</p>
    <div class='summary'>
        <div class='card'><h3>€{total:,.2f}</h3><p>Total Ventas</p></div>
        <div class='card'><h3>{unidades:,}</h3><p>Unidades</p></div>
        <div class='card'><h3>{prods}</h3><p>Productos</p></div>
    </div>
    <table><thead><tr><th>Fecha</th><th>Producto</th><th>Unidades</th><th>Total</th></tr></thead><tbody>{filas}</tbody></table>
</div></body></html>"""
    return html

informe = generar_informe(datos, "Informe de Ventas Semanal")
with open('proyecto/informes/informe.html', 'w', encoding='utf-8') as f:
    f.write(informe)
print("✓ Informe generado: proyecto/informes/informe.html")
```

## Paso 4: Envío por Correo (Simulado)

```python
def enviar_correo(destinatario):
    print(f"=== Envío de Correo ===")
    print(f"Para: {destinatario}")
    print(f"Asunto: Informe de Ventas - {datetime.now().strftime('%d/%m/%Y')}")
    print(f"Estado: ✓ Enviado")
    return True

enviar_correo('equipo@fortia.com')
```

## Paso 5: Compresión y Archivado

```python
import zipfile
from datetime import datetime

def comprimir(ruta_informe, destino):
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    zip_path = os.path.join(destino, f'informe_{ts}.zip')
    with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_DEFLATED) as z:
        z.write(ruta_informe, 'informe.html')
    return zip_path, os.path.getsize(zip_path)

zip_path, tamano = comprimir('proyecto/informes/informe.html', 'proyecto/respaldos')
print(f"✓ Comprimido: {zip_path} ({tamano:,} bytes)")
```

## Clase Integradora

```python
class SistemaReportes:
    def __init__(self, base='proyecto'):
        self.base = base
        self.datos = None
        self.ts = datetime.now().strftime("%Y%m%d_%H%M%S")
        for s in ['datos', 'informes', 'respaldos']:
            os.makedirs(os.path.join(base, s), exist_ok=True)
    
    def generar_datos(self, dias=7):
        self.datos = generar_datos(dias)
        return len(self.datos)
    
    def crear_informe(self, titulo):
        html = generar_informe(self.datos, titulo)
        ruta = os.path.join(self.base, 'informes', f'informe_{self.ts}.html')
        with open(ruta, 'w', encoding='utf-8') as f: f.write(html)
        return ruta
    
    def enviar_correo(self, destinatario):
        enviar_correo(destinatario)
        return True
    
    def comprimir(self):
        ruta = os.path.join(self.base, 'informes', f'informe_{self.ts}.html')
        zip_path, tamano = comprimir(ruta, os.path.join(self.base, 'respaldos'))
        return zip_path, tamano
    
    def ejecutar(self, destinatario):
        print("=" * 50)
        print("SISTEMA DE REPORTES AUTOMATIZADO")
        print("=" * 50)
        n = self.generar_datos()
        print(f"[1/4] ✓ Datos generados: {n} registros")
        ruta = self.crear_informe("Informe FORTIA")
        print(f"[2/4] ✓ Informe: {os.path.basename(ruta)}")
        self.enviar_correo(destinatario)
        print(f"[3/4] ✓ Correo enviado")
        zip_path, tamano = self.comprimir()
        print(f"[4/4] ✓ Comprimido: {os.path.basename(zip_path)}")
        print("=" * 50)
        return True

sistema = SistemaReportes()
sistema.ejecutar('equipo@fortia.com')
```
