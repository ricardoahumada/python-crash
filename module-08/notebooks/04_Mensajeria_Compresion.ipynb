## 4.2 Compresión de Archivos con zipfile

```python
# Compresión con zipfile
import zipfile
import os

os.makedirs('datos/comprimir', exist_ok=True)
for i in range(1, 4):
    with open(f'datos/comprimir/archivo_{i}.txt', 'w') as f:
        f.write(f'Contenido {i}\n')

def comprimir_carpeta(carpeta_origen, archivo_zip):
    with zipfile.ZipFile(archivo_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for root, dirs, files in os.walk(carpeta_origen):
            for file in files:
                ruta = os.path.join(root, file)
                arcname = os.path.relpath(ruta, carpeta_origen)
                zipf.write(ruta, arcname)
    return archivo_zip

print("=== Compresión ZIP ===")
zip_path = comprimir_carpeta('datos/comprimir', 'datos/respaldos.zip')
print(f"✓ Creado: {zip_path}")

with zipfile.ZipFile(zip_path, 'r') as z:
    for info in z.infolist():
        print(f"  - {info.filename}")
```

## 4.3 Compresión con tarfile

```python
import tarfile

def comprimir_tar_gz(carpeta, salida):
    with tarfile.open(salida, 'w:gz') as tar:
        tar.add(carpeta, arcname=os.path.basename(carpeta))
    return salida

print("=== Compresión tar.gz ===")
tar_path = comprimir_tar_gz('datos/comprimir', 'datos/respaldos.tar.gz')
print(f"✓ Creado: {tar_path}")
```

## Práctica: Sistema de Informe + Correo + Compresión

```python
# PRÁCTICA: Sistema Completo
from datetime import datetime
import zipfile
import os

def sistema_completo(datos, carpeta_salida):
    os.makedirs(carpeta_salida, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    
    # 1. Generar informe
    html = f"<html><body><h1>Informe</h1><p>Generado: {ts}</p>"
    for d in datos:
        html += f"<tr><td>{d['concepto']}</td><td>{d['valor']}</td></tr>"
    html += "</body></html>"
    
    # 2. Guardar
    html_file = os.path.join(carpeta_salida, f'informe_{ts}.html')
    with open(html_file, 'w') as f: f.write(html)
    print(f"1. ✓ Informe: {html_file}")
    
    # 3. Simular correo
    print(f"2. ✓ Correo enviado a destinatario@ejemplo.com")
    
    # 4. Comprimir
    zip_file = os.path.join(carpeta_salida, f'informe_{ts}.zip')
    with zipfile.ZipFile(zip_file, 'w') as z:
        z.write(html_file, os.path.basename(html_file))
    print(f"3. ✓ Comprimido: {zip_file}")
    
    return {'html': html_file, 'zip': zip_file}

print("=== PRÁCTICA: Sistema Completo ===")
datos = [
    {'concepto': 'Ventas', 'valor': '€45,230'},
    {'concepto': 'Clientes', 'valor': '127'}
]
resultado = sistema_completo(datos, 'datos/salida')
print("\nCompletado!")
```
