## 1.1 Principios de Automatización

La automatización consiste en convertir tareas manuales repetitivas en procesos ejecutables sin intervención humana. Antes de automatizar, debemos:

1. **Identificar** tareas susceptibles de automatización
2. **Evaluar** costos y beneficios
3. **Diseñar** la solución
4. **Implementar** y probar

### ¿Qué tareas se benefician de la automatización?

- Tareas repetitivas que se ejecutan con frecuencia
- Procesos con reglas claras y definidas
- Operaciones propensas a errores humanos
- Trabajos que consumen tiempo significativo

## 1.2 Scripting Básico para Automatización

Un script de automatización sigue el patrón **Entrada → Proceso → Salida**:

```python
# Ejemplo: Patrón básico de automatización
import os
from datetime import datetime

def respaldar_archivos(carpeta_origen, carpeta_destino):
    """Copia archivos de origen a destino con marca de tiempo"""
    if not os.path.exists(carpeta_destino):
        os.makedirs(carpeta_destino)
    
    archivos = os.listdir(carpeta_origen)
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    
    for archivo in archivos:
        ruta_origen = os.path.join(carpeta_origen, archivo)
        if os.path.isfile(ruta_origen):
            nombre_nuevo = f"{timestamp}_{archivo}"
            ruta_destino = os.path.join(carpeta_destino, nombre_nuevo)
            print(f"Copiado: {archivo} -> {nombre_nuevo}")
    
    return len(archivos)

# Ejecutar
print("=== Sistema de Respaldo Automatizado ===")
count = respaldar_archivos("datos/origen", "datos/respaldo")
print(f"Total archivos procesados: {count}")
```

## Práctica: Crear un Script de Respaldo Automatizado

**Objetivo**: Crear un script que respalde automáticamente archivos de una carpeta a otra.

### Pasos:

1. Importar módulos necesarios
2. Definir función de respaldo con marca de tiempo
3. Añadir manejo de errores
4. Probar con carpetas de prueba

```python
# PRÁCTICA: Script de Respaldo Completo
import os
import shutil
from datetime import datetime

class SistemaRespaldo:
    def __init__(self, origen, destino):
        self.origen = origen
        self.destino = destino
        self.timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    
    def respaldar(self):
        """Ejecuta el respaldo de archivos"""
        if not os.path.exists(self.origen):
            print(f"Error: La carpeta {self.origen} no existe")
            return 0
        
        if not os.path.exists(self.destino):
            os.makedirs(self.destino)
            print(f"Carpeta destino creada: {self.destino}")
        
        archivos = [f for f in os.listdir(self.origen) 
                   if os.path.isfile(os.path.join(self.origen, f))]
        
        count = 0
        for archivo in archivos:
            origen_path = os.path.join(self.origen, archivo)
            nombre_timestamp = f"{self.timestamp}_{archivo}"
            destino_path = os.path.join(self.destino, nombre_timestamp)
            
            try:
                shutil.copy2(origen_path, destino_path)
                print(f"✓ {archivo} -> {nombre_timestamp}")
                count += 1
            except Exception as e:
                print(f"✗ Error al copiar {archivo}: {e}")
        
        return count

# Ejecutar práctica
print("=== PRÁCTICA: Script de Respaldo ===")
print("Creando carpetas de prueba...")

# Crear carpetas de prueba
os.makedirs("datos/origen", exist_ok=True)
os.makedirs("datos/respaldo", exist_ok=True)

# Crear archivos de prueba
for i in range(1, 4):
    with open(f"datos/origen/archivo_{i}.txt", "w") as f:
        f.write(f"Contenido del archivo {i}\nFecha: {datetime.now()}")

print(f"Archivos creados en datos/origen")

# Ejecutar respaldo
sistema = SistemaRespaldo("datos/origen", "datos/respaldo")
total = sistema.respaldar()
print(f"\nTotal archivos respaldados: {total}")
```
