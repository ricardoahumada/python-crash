## 3.1 Creación de Tablas HTML

Las tablas HTML son fundamentales para mostrar datos de forma estructurada.

```python
# Generación de tablas HTML con Python
def crear_tabla_html(datos, columnas):
    """Genera una tabla HTML desde una lista de diccionarios"""
    
    headers_html = "<tr>" + "".join(f"<th>{col['titulo']}</th>" for col in columnas) + "</tr>"
    
    rows_html = ""
    for item in datos:
        row = "<tr>" + "".join(f"<td>{item.get(col['clave'], '')}</td>" for col in columnas) + "</tr>"
        rows_html += row
    
    tabla = f"""
<table class="data-table">
    <thead>
        {headers_html}
    </thead>
    <tbody>
        {rows_html}
    </tbody>
</table>
    """
    return tabla

empleados = [
    {'nombre': 'Ana García', 'departamento': 'Ventas', 'salario': 45000},
    {'nombre': 'Carlos López', 'departamento': 'Marketing', 'salario': 42000},
    {'nombre': 'María Rodríguez', 'departamento': 'IT', 'salario': 52000},
    {'nombre': 'Juan Martínez', 'departamento': 'Ventas', 'salario': 38000}
]

columnas = [
    {'titulo': 'Nombre', 'clave': 'nombre'},
    {'titulo': 'Departamento', 'clave': 'departamento'},
    {'titulo': 'Salario', 'clave': 'salario'}
]

tabla = crear_tabla_html(empleados, columnas)
print(tabla)
```

## 3.2 Estilos CSS para Tablas Profesionales

## Práctica: Crear una Tabla HTML con Formato CSS Profesional

```python
# PRÁCTICA: Tabla de Calificaciones
estudiantes = [
    {'nombre': 'Pablo Fernández', 'nota': 8.5, 'asistencia': '92%'},
    {'nombre': 'Laura Gutiérrez', 'nota': 9.2, 'asistencia': '98%'},
    {'nombre': 'Diego Ramírez', 'nota': 7.3, 'asistencia': '85%'},
    {'nombre': 'Sofia Hernández', 'nota': 8.8, 'asistencia': '95%'},
    {'nombre': 'Martín López', 'nota': 6.5, 'asistencia': '78%'},
]

def generar_tabla_calificaciones(datos, titulo):
    filas = ""
    for est in datos:
        color_nota = '#27ae60' if est['nota'] >= 8 else ('#f39c12' if est['nota'] >= 6 else '#e74c3c')
        filas += f"<tr><td>{est['nombre']}</td><td style='text-align: center; color: {color_nota};'>{est['nota']}</td><td style='text-align: center;'>{est['asistencia']}</td></tr>"
    
    promedio = sum(e['nota'] for e in datos) / len(datos)
    
    html = f"""
<!DOCTYPE html>
<html><head><style>
    body {font-family: 'Segoe UI', Arial, sans-serif; margin: 40px;}
    h1 {color: #2c3e50; text-align: center;}
    table {width: 100%; border-collapse: collapse; margin-top: 20px;}
    th {background-color: #3498db; color: white; padding: 15px; text-align: left;}
    td {padding: 12px 15px; border-bottom: 1px solid #ecf0f1;}
    tr:hover {background-color: #f8f9fa;}
    .resumen {background: #ecf0f1; padding: 15px; border-radius: 8px; margin-top: 20px; text-align: center;}
</style></head>
<body>
    <h1>{titulo}</h1>
    <table>
        <thead><tr><th>Estudiante</th><th style='text-align: center;'>Nota</th><th style='text-align: center;'>Asistencia</th></tr></thead>
        <tbody>{filas}</tbody>
    </table>
    <div class='resumen'><strong>Nota media:</strong> {promedio:.2f}</div>
</body></html>
    """
    return html

print("=== PRÁCTICA: Tabla de Calificaciones ===")
html = generar_tabla_calificaciones(estudiantes, "Calificaciones del Módulo 8")
with open('tabla_calificaciones.html', 'w', encoding='utf-8') as f:
    f.write(html)
print("✓ Tabla generada: tabla_calificaciones.html")
```
