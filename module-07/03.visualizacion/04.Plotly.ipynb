# Plotly



https://plotly.com/python/

Plotly es una biblioteca de código abierto para crear visualizaciones interactivas de alta calidad. 

Plotly Python (Plotly.py) es una biblioteca de visualización interactiva y publicación de calidad científica que fue desarrollada por la empresa Plotly. Se puede usar para crear una amplia variedad de gráficos, desde simples gráficos de líneas hasta complejos dashboards 3D.

Con Plotly, puedes crear visualizaciones hermosas e interactivas que se pueden mostrar en navegadores web, notebooks Jupyter, o exportar como archivos HTML estáticos o imágenes.

Plotly supports:
- **Gráficos 2D/3D**: líneas, barras, dispersión, área, burbujas
- **Mapas**: coropletas, mapas de calor, mapas de puntos
- **Gráficos estadísticos**: histogramas, boxplots, violin plots
- **Gráficos científicos**: contornos, heatmaps, dendrogramas
- **Dashboards**: combine múltiples gráficos con componentes interactivos

## Plotly en Notebooks

Plotly funciona directamente en Jupyter Notebooks sin ninguna configuración adicional. 
Solo necesitas importar la biblioteca y comenzar a crear visualizaciones.

La sintaxis básica sigue el patrón:
1. Crear una figura con `go.Figure()` o `px.plot()`
2. Agregar trazas (traces) con métodos como `add_trace()`
3. Mostrar la figura con `fig.show()`

```python
import plotly.graph_objects as go
import plotly.express as px
import numpy as np

# Verificar que plotly está instalado
print(f"Plotly version: {px.__version__}")
```

## Interfaces de Plotly

Plotly ofrece dos interfaces principales para crear gráficos:

### 1. Plotly Express (`plotly.express` o `px`)
- API simple y fácil de usar
- Ideal para gráficos comunes con mínima configuración
- Abreviada: `px.scatter()`, `px.bar()`, `px.line()`, etc.

### 2. Graph Objects (`plotly.graph_objects` o `go`)
- API de bajo nivel con control total
- Ideal para personalizaciones avanzadas y gráficos complejos
- Más detallado: `go.Scatter()`, `go.Bar()`, `go.Figure()`, etc.

```python
# Ejemplo con Plotly Express (px)
import plotly.express as px

df = px.data.iris()  # dataset de ejemplo incluido en plotly
df.head()
```

```python
# Crear un scatter plot simple con px
fig = px.scatter(df, x="sepal_width", y="sepal_length", title="Iris Dataset - Sepal Dimensions")
fig.show()
```

```python
# Mismo gráfico con Graph Objects (go)
import plotly.graph_objects as go

fig = go.Figure(data=go.Scatter(x=df["sepal_width"], y=df["sepal_length"], mode="markers"))
fig.update_layout(title="Iris Dataset - Sepal Dimensions (go)")
fig.show()
```

## Componentes de Plotly

Los gráficos de Plotly se componen de:

- **Data (Trazados)**: Representaciones visuales de datos (scatter, bar, line, etc.)
- **Layout**: Elementos no relacionados con datos (título, leyenda, ejes, fondo, etc.)
- **Figure**: El objeto principal que combina Data y Layout



## Plotting Básico

### Gráfico de líneas

```python
# Datos básicos
x = [1, 2, 3, 4, 5]
y = [6, 7, 2, 1, 5]

# Usar px.line() para crear un gráfico de líneas
fig = px.line(x=x, y=y, title="Basic Line Plot")
fig.show()
```

```python
# Personalizar el gráfico con go
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=x, 
    y=y, 
    mode="lines+markers",  # muestra línea y puntos
    name="Temperature",
    line=dict(width=2, color="blue"),
    marker=dict(size=8, color="red")
))

fig.update_layout(
    title="Customized Line Plot",
    xaxis_title="x",
    yaxis_title="y",
    height=400,
    width=600
)

fig.show()
```

### Diagrama de dispersión (Scatter)

```python
# Crear un scatter plot
x = [1, 2, 3, 4, 5]
y = [6, 7, 2, 4, 5]

fig = px.scatter(x=x, y=y, title="Basic Scatter Plot")
fig.show()
```

```python
# Datos categóricos en gráficos de barras
fruits = ["Apples", "Pears", "Nectarines", "Plums", "Grapes", "Strawberries"]
counts = [5, 3, 4, 2, 4, 6]

fig = px.bar(x=fruits, y=counts, title="Fruit Counts", color=fruits)
fig.show()
```

```python
# Barra horizontal
fig = px.bar(x=counts, y=fruits, orientation="h", title="Fruit Counts (Horizontal)")
fig.show()
```

### Gráficos de pastel (Pie)

```python
# Gráfico de pastel
fig = px.pie(
    values=counts, 
    names=fruits, 
    title="Fruit Distribution",
    hole=0.3  # dona en lugar de pastel
)
fig.show()
```

### Histogramas

```python
# Generar datos aleatorios
np.random.seed(42)
data = np.random.randn(1000)

fig = px.histogram(
    data, 
    nbins=30, 
    title="Histogram of Normal Distribution",
    labels={"value": "Value", "count": "Frequency"}
)
fig.update_layout(bargap=0.1)
fig.show()
```

### Box Plots

```python
# Box plot con px
df_tips = px.data.tips()
fig = px.box(
    df_tips, 
    x="day", 
    y="total_bill", 
    color="sex",
    title="Total Bill by Day"
)
fig.show()
```

### Gráficos de área

```python
# Área simple
x = [1, 2, 3, 4, 5]
y = [6, 7, 2, 1, 5]

fig = px.area(x=x, y=y, title="Area Chart")
fig.show()
```

## Múltiples trazados

```python
# Crear un gráfico con múltiples trazados
x = [1, 2, 3, 4, 5]
y1 = [6, 7, 2, 1, 5]
y2 = [5, 4, 3, 2, 1]
y3 = [1, 3, 5, 7, 9]

fig = go.Figure()

# Agregar trazas
fig.add_trace(go.Scatter(x=x, y=y1, mode="lines+markers", name="Series 1"))
fig.add_trace(go.Scatter(x=x, y=y2, mode="lines+markers", name="Series 2"))
fig.add_trace(go.Scatter(x=x, y=y3, mode="lines+markers", name="Series 3"))

fig.update_layout(title="Multiple Series", xaxis_title="x", yaxis_title="y")
fig.show()
```

```python
# Subplots con make_subplots
from plotly.subplots import make_subplots

fig = make_subplots(rows=1, cols=2, subplot_titles=("Series 1", "Series 2"))

fig.add_trace(go.Scatter(x=x, y=y1, mode="lines+markers"), row=1, col=1)
fig.add_trace(go.Scatter(x=x, y=y2, mode="lines+markers"), row=1, col=2)

fig.update_layout(height=400, width=800, title_text="Side by Side")
fig.show()
```

## Personalización de Estilos

### Colores en Plotly

Plotly ofrece varias formas de especificar colores:

1. **Colores CSS/named**: "red", "blue", "green"
2. **Hexadecimal**: "#FF5733", "#3366FF"
3. **RGB/RGBA**: "rgb(255, 87, 51)", "rgba(255, 87, 51, 0.5)"
4. **Paletas de Plotly**: "plotly", "plotlywhite", "earth", "viridis"

```python
# Usar diferentes formatos de color
fig = go.Figure()

colors = ["red", "#FF5733", "rgb(51, 102, 255)", "rgba(100, 200, 100, 0.5)"]
y_values = [[6, 7, 2], [5, 4, 3], [3, 5, 7], [2, 8, 4]]

for i, (color, y) in enumerate(zip(colors, y_values)):
    fig.add_trace(go.Bar(x=["A", "B", "C"], y=y, name=f"Color {i+1}", marker_color=color))

fig.update_layout(title="Different Color Formats")
fig.show()
```

```python
# Usar paletas de colores continuas
fig = px.scatter(
    x=np.random.randn(100),
    y=np.random.randn(100),
    color=np.random.randn(100),
    color_continuous_scale="viridis",
    title="Continuous Color Scale"
)
fig.show()
```

### Personalizar ejes

```python
# Personalizar ejes
x = list(range(1, 50))
y = [pow(i, 2) for i in x]

fig = px.scatter(x=x, y=y, title="Squared Values")

fig.update_xaxes(
    title="X Axis",
    showgrid=True,
    gridwidth=1,
    gridcolor="lightgray",
    zeroline=True,
    zerolinewidth=2,
    zerolinecolor="red",
    tickangle=45
)

fig.update_yaxes(
    title="Y Axis (X²)",
    showgrid=True,
    gridwidth=1,
    gridcolor="lightblue",
    scaleanchor="x",
    scaleratio=1
)

fig.show()
```

### Personalizar título y leyenda

```python
# Personalizar título
fig = go.Figure(data=go.Scatter(x=[1, 2, 3], y=[4, 5, 6]))

fig.update_layout(
    title=dict(
        text="<b>Main Title</b><br><sup>Subtitle goes here</sup>",
        x=0.5,  # centrar
        xanchor="center",
        y=0.95,
        yanchor="top",
        font=dict(size=24, color="darkblue")
    ),
    legend=dict(
        title="Legend Title",
        orientation="h",
        yanchor="bottom",
        y=-0.3,
        xanchor="center",
        x=0.5,
        bgcolor="lightgray",
        bordercolor="black",
        borderwidth=1
    )
)

fig.show()
```

### Plantillas (Themes)

Plotly incluye varias plantillas predefinidas:
- `plotly` (default)
- `plotly_white`
- `plotly_dark`
- `ggplot2`
- `seaborn`
- `simple_white`

```python
# Usar diferentes plantillas
import plotly.io as pio

# Crear figura base
x = [1, 2, 3, 4, 5]
y = [6, 7, 2, 4, 5]
fig = go.Figure(data=go.Scatter(x=x, y=y, mode="markers"))

# Probar diferentes plantillas
templates = ["plotly", "plotly_white", "ggplot2", "seaborn"]

for template in templates:
    fig_temp = go.Figure(data=go.Scatter(x=x, y=y, mode="markers", name=template))
    fig_temp.update_layout(title=f"Template: {template}", template=template, height=300)
    fig_temp.show()
```

## Anotaciones

### Añadir texto y flechas

```python
# Añadir anotaciones
x = list(range(1, 50))
y = [pow(i, 2) for i in x]

fig = px.scatter(x=x, y=y, title="Annotation Example")

# Añadir anotación
fig.add_annotation(
    x=5,
    y=25,
    text="Start Point",
    showarrow=True,
    arrowhead=2,
    arrowsize=1,
    arrowwidth=2,
    arrowcolor="red",
    ax=40,
    ay=-30,
    font=dict(size=14, color="red"),
    bgcolor="yellow",
    bordercolor="red",
    borderwidth=1,
    borderpad=4
)

# Añadir anotación sin flecha
fig.add_annotation(
    x=40,
    y=1600,
    text="Maximum Area",
    showarrow=False,
    font=dict(size=16, color="green"),
    bgcolor="lightgreen"
)

fig.show()
```

### Formas (Shapes)

```python
# Añadir formas al gráfico
x = list(range(1, 50))
y = [pow(i, 2) for i in x]

fig = px.scatter(x=x, y=y, title="Shapes Example")

# Línea horizontal
fig.add_shape(
    type="line",
    x0=1,
    x1=49,
    y0=400,
    y1=400,
    line=dict(color="red", width=2, dash="dash")
)

# Rectángulo
fig.add_shape(
    type="rect",
    x0=10,
    x1=20,
    y0=100,
    y1=400,
    fillcolor="lightblue",
    opacity=0.3,
    line=dict(color="blue", width=2)
)

fig.show()
```

## Layouts y Subplots

```python
# Grid specs para layouts complejos
from plotly.subplots import make_subplots

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=("Plot 1", "Plot 2", "Plot 3", "Plot 4"),
    specs=[[{"type": "scatter"}, {"type": "bar"}],
           [{"type": "pie"}, {"type": "box"}]]
)

# Scatter
fig.add_trace(go.Scatter(x=[1, 2, 3], y=[4, 5, 6], mode="markers"), row=1, col=1)

# Bar
fig.add_trace(go.Bar(x=["A", "B", "C"], y=[3, 6, 9]), row=1, col=2)

# Pie
fig.add_trace(go.Pie(values=[30, 40, 30], labels=["X", "Y", "Z"]), row=2, col=1)

# Box
fig.add_trace(go.Box(y=[1, 2, 3, 5, 7, 8, 9]), row=2, col=2)

fig.update_layout(height=600, width=800, title_text="4-Plot Grid")
fig.show()
```

## Interactividad

Una de las grandes ventajas de Plotly es la interactividad automática de los gráficos.

### Herramientas integradas

Por defecto, los gráficos Plotly incluyen:
- **Zoom**: Selecciona un área para hacer zoom
- **Pan**: Desplaza el gráfico
- **Zoom in/out**: Rueda del ratón
- **Reset**: Vuelve al estado original
- **Hover**: Muestra información al pasar el cursor
- **Download**: Guarda como imagen PNG

```python
# tooltips (hovertemplate)
df = px.data.iris()

fig = px.scatter(
    df, 
    x="sepal_width", 
    y="sepal_length",
    color="species",
    hover_data=["petal_length", "petal_width"],
    title="Iris - Hover for Details"
)

fig.update_traces(
    hovertemplate="<b>%{fullData.name}</b><br>" +
                  "Sepal Width: %{x}<br>" +
                  "Sepal Length: %{y}<br>" +
                  "Petal Length: %{customdata[0]}<br>" +
                  "Petal Width: %{customdata[1]}<br>" +
                  "<extra></extra>"
)

fig.show()
```

### Botones de control

```python
# Botones para cambiar entre tipos de gráficos
df = px.data.tips()

fig = go.Figure()

# Añadir traces ocultos inicialmente
fig.add_trace(go.Scatter(x=df["total_bill"], y=df["tip"], mode="markers", name="Scatter", visible=True))
fig.add_trace(go.Histogram(x=df["total_bill"], name="Histogram", visible=False))

# Crear botones
fig.update_layout(
    updatemenus=[
        dict(
            type="buttons",
            direction="right",
            x=0.5,
            y=1.15,
            xanchor="center",
            buttons=[
                dict(label="Scatter",
                     method="update",
                     args=[{"visible": [True, False]},
                           {"title": "Tips - Scatter"}]),
                dict(label="Histogram",
                     method="update",
                     args=[{"visible": [False, True]},
                           {"title": "Tips - Histogram"}])
            ]
        )
    ]
)

fig.show()
```

### Sliders y Dropdowns

```python
# Crear un slider para cambiar datos
np.random.seed(42)
x = list(range(20))

fig = go.Figure()

# Crear traces para diferentes valores de n
for n in [2, 3, 4]:
    y = [pow(i, n) for i in x]
    fig.add_trace(go.Scatter(x=x, y=y, mode="lines", name=f"x^{n}", visible=False))

# Mostrar solo la primera trace
fig.data[0].visible = True

# Crear steps para el slider
steps = []
for i in range(len(fig.data)):
    step = dict(
        method="update",
        args=[{"visible": [j == i for j in range(len(fig.data))]},
              {"title": f"Polynomial: x^{i+2}"}]
    )
    steps.append(step)

# Crear slider
sliders = [dict(
    active=0,
    currentvalue={"prefix": "Power: "},
    pad={"t": 50},
    steps=steps
)]

fig.update_layout(
    sliders=sliders,
    title="Polynomial Curves"
)

fig.show()
```

## Animaciones

## Ejemplo integrador

```python
# Dashboard completo combinando múltiples elementos

# Preparar datos
df_tips = px.data.tips()

# Crear figura con subplots
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=("Total Bill Distribution", "Tips by Day", "Bill vs Tip", "Tips by Gender"),
    specs=[[{"type": "histogram"}, {"type": "bar"}],
           [{"type": "scatter"}, {"type": "box"}]]
)

# 1. Histograma de total_bill
fig.add_trace(go.Histogram(x=df_tips["total_bill"], name="Total Bill", nbinsx=20), row=1, col=1)

# 2. Bar chart de tips por día
tips_by_day = df_tips.groupby("day")["tip"].sum().reset_index()
fig.add_trace(go.Bar(x=tips_by_day["day"], y=tips_by_day["tip"], name="Tips by Day"), row=1, col=2)

# 3. Scatter de bill vs tip
fig.add_trace(go.Scatter(
    x=df_tips["total_bill"],
    y=df_tips["tip"],
    mode="markers",
    marker=dict(
        color=df_tips["size"],
        colorscale="Viridis",
        showscale=True
    ),
    name="Bill vs Tip"
), row=2, col=1)

# 4. Box plot por género
fig.add_trace(go.Box(x=df_tips["sex"], y=df_tips["tip"], name="Tips by Gender"), row=2, col=2)

# Actualizar layout
fig.update_layout(
    height=700,
    width=900,
    title=dict(text="<b>Tips Dashboard</b>", x=0.5),
    showlegend=False
)

# Actualizar títulos de ejes
fig.update_xaxes(title_text="Total Bill ($)", row=1, col=1)
fig.update_yaxes(title_text="Frequency", row=1, col=1)
fig.update_xaxes(title_text="Day", row=1, col=2)
fig.update_yaxes(title_text="Total Tips ($)", row=1, col=2)
fig.update_xaxes(title_text="Total Bill ($)", row=2, col=1)
fig.update_yaxes(title_text="Tip ($)", row=2, col=1)
fig.update_xaxes(title_text="Gender", row=2, col=2)
fig.update_yaxes(title_text="Tip ($)", row=2, col=2)

fig.show()
```
